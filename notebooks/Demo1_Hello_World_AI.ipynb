# Demo 1 · The Agentic Shift: your first AI-built web page 🚀
**Idea of today:** you are the **Director**. You say *what* you want in plain English. The AI does the typing. You check the result and ask for changes.

| Old way | New way (Director) |
|---|---|
| Learn the code, type every line | Describe the result, AI writes the code |
| Hours | Minutes |

**What you will do:** connect to an AI → say hello → get the AI to build a web page → change it → save it.

## ⚙️ Setup (run once)
Click the ▶ button on the left of the cell below. When a box appears, **paste your Groq API key** and press **Enter**.

> 🔑 **No key yet?** Go to [console.groq.com/keys](https://console.groq.com/keys) → sign in with Google → **Create API Key** → copy it.
> Treat it like a password. Do not share it or put it in slides.

````
# This cell installs the AI library and connects to Groq. You do not need to understand it.
%pip install -q groq
import re, json, html
from getpass import getpass
from groq import Groq
from IPython.display import HTML, display, Markdown

client = Groq(api_key=getpass("Paste your Groq API key and press Enter: "))
MODEL = "openai/gpt-oss-20b"   # fast, free-tier friendly. You can also try "openai/gpt-oss-120b"

def ask_ai(prompt, system="You are a friendly, helpful assistant."):
    """Send a message to the AI and get its answer back as text."""
    reply = client.chat.completions.create(
        model=MODEL,
        messages=[{"role": "system", "content": system}, {"role": "user", "content": prompt}],
    )
    return reply.choices[0].message.content

def clean_code(text):
    """AI often wraps code in ``` marks. This keeps only the code inside."""
    m = re.search(r"```(?:[a-zA-Z]*)\s*(.*?)```", text, re.S)
    return (m.group(1) if m else text).strip()

def show_page(page, height=480):
    """Show a web page right here inside the notebook."""
    display(HTML(f'<iframe srcdoc="{html.escape(page)}" width="100%" height="{height}" '
                 f'style="border:1px solid #ccc;border-radius:8px"></iframe>'))

print(ask_ai("Reply with exactly: ✅ Connected! The AI is ready.", system="Follow the instruction exactly."))
````

## Step 1 · Say hello to the AI 👋
Run the cell. You just sent a message to an AI model and it answered. That is all an "API call" is: a message and a reply.

````
print(ask_ai("Say hello to a class of first-time coders in one fun, encouraging sentence."))
````

## Step 2 · Be the Director: describe a web page 🎬
Read the request below. It is just English. Run it and watch the AI build the page.

````
DESIGNER = "You are an expert web designer. Reply with ONLY one complete HTML file (CSS and JavaScript inside it). No explanations."

request = """
Create a simple, colourful 'Hello World' web page.
- A big heading: Hello World!
- A line under it: Built by directing AI, not by typing code.
- A button that says 'Click me' and shows a friendly message when clicked.
- Centred on the screen, modern and clean.
"""

page = clean_code(ask_ai(request, system=DESIGNER))
show_page(page)
````

✅ **You wrote zero lines of code.** Click the button in the page above. It works!

👀 *Curious?* Run the next cell to peek at the code the AI wrote. You do not need to understand it.

````
print(page[:1200], "\n...")
````

## Step 3 · Change your mind (this is the real skill) 🔁
Directors give feedback. Edit the text inside `change` (put your own name!) and run the cell.

````
change = "Make the background orange, add 'Made by Your Name' under the heading, and add a fun emoji to the title."

page = clean_code(ask_ai(f"Here is a web page:\n{page}\n\nChange it like this: {change}\nReturn the full updated HTML.", system=DESIGNER))
show_page(page)
````

## Step 4 · Save your page 💾
This saves `hello.html`. In Colab it also downloads to your laptop. Double-click the file to open it in your browser.

````
open("hello.html", "w", encoding="utf-8").write(page)
print("Saved hello.html")
try:
    from google.colab import files; files.download("hello.html")
except ImportError:
    pass  # not in Colab: the file is in the same folder as this notebook
````
